```python
# Problem4:Data Loading and Exploration
import pandas as pd

# Creating customer dataset
data = {
    'Customer_ID': [101, 102, 103, 104, 105,
                    106, 107, 108, 109, 110],

    'Name': ['Asha', 'Rohan', 'Neha', 'Arjun', 'Priya',
             'Karan', 'Meera', 'Dev', 'Anjali', 'Vivek'],

    'Age': [22, 35, 28, None, 31, 26, 39, 45, 29, 33],

    'Gender': ['Female', 'Male', 'Female', 'Male', 'Female',
               'Male', 'Female', 'Male', 'Female', 'Male'],

    'Purchase_Amount': [5000, 8000, 4500, 12000, 7000,
                        6000, 9000, 15000, 5500, 10000]
}

# Loading data into DataFrame
df = pd.DataFrame(data)

# Display first 5 rows
print("First 5 Rows:")
print(df.head())

# Display summary statistics
print("\nSummary Statistics:")
print(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Handle missing values
df['Age'] = df['Age'].fillna(df['Age'].median())

# Display updated DataFrame
print("\nData After Handling Missing Values:")
print(df)
```

```python
#Problem6: Data Filtering and Selection
import pandas as pd

data = {
    'Customer_ID': [101, 102, 103, 104, 105,
                    106, 107, 108, 109, 110],

    'Name': ['Asha', 'Rohan', 'Neha', 'Arjun', 'Priya',
             'Karan', 'Meera', 'Dev', 'Anjali', 'Vivek'],

    'Age': [22, 35, 28, None, 31, 26, 39, 45, 29, 33],

    'Gender': ['Female', 'Male', 'Female', 'Male', 'Female',
               'Male', 'Female', 'Male', 'Female', 'Male'],

    'Purchase_Amount': [5000, 8000, 4500, 12000, 7000,
                        6000, 9000, 15000, 5500, 10000]
}

df = pd.DataFrame(data)

# Handling missing Age values
df['Age'] = df['Age'].fillna(df['Age'].median())

# 1. Add Discounted_Price column
df['Discounted_Price'] = df['Purchase_Amount'] * 0.90

print("Data After Adding Discount:")
print(df)

# 2. Calculate average Purchase Amount by Gender
print("\nAverage Purchase Amount by Gender:")
print(df.groupby('Gender')['Purchase_Amount'].mean())

# 3. Sort Purchase Amount in descending order
print("\nData Sorted by Purchase Amount:")
print(df.sort_values(by='Purchase_Amount', ascending=False))
```
