```python
#Given a filename and a list of strings, create a file containing pairs of strings joined together, with each pair on a separate line.
def zipper(filename, strings):
    with open(filename, 'w') as file:
        for i in range(0, len(strings), 2):
            file.write(strings[i] + strings[i+1] + '\n')
# Example
strings = ['Hello', 'World', 'Python', 'Programming',
           'Data', 'Science']

zipper('output.txt', strings)

print("File created successfully!")

```

```python
#Convert dates from MM/DD/YYYY format into English sentences.
def reformat_date(dates):

    months = {
        '01': 'January',
        '02': 'February',
        '03': 'March',
        '04': 'April',
        '05': 'May',
        '06': 'June',
        '07': 'July',
        '08': 'August',
        '09': 'September',
        '10': 'October',
        '11': 'November',
        '12': 'December'
    }

    result = []

    for date in dates:
        month, day, year = date.split('/')

        sentence = "Office Hours on " + months[month] + " " + day + ", " + year

        result.append(sentence)

    return result
    
    
    # Example
dates = [
    "01/15/2025",
    "03/09/2025",
    "07/21/2025",
    "11/05/2025"
]

print(reformat_date(dates))


```
