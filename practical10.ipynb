```python
#1.Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset('iris')
print(df.head())
```

```python
#2. Scatter plot
plt.figure(figsize=(6,4))
sns.scatterplot(
    data=df,
    x='sepal_length',
    y='sepal_width',
    hue='species',
    style='species',
    s=60
)

plt.title('Scatter Plot of Iris Dataset')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()

```

Analysis:
1. Scatter plot shows the relationship between sepal length and sepal width.
2. Setosa flowers are clearly separated from the other two species.
3. Versicolor and Virginica show some overlapping points.
4. Setosa generally has smaller sepal length but greater sepal width.
5. The plot helps to identify patterns and differences among the three flower species.

```python
#3. Density Plot
plt.figure(figsize=(4,2))
sns.kdeplot(
    data=df,
    x='petal_length',
    hue='species',
    fill=True,
    alpha=0.8
)

plt.title('Density Plot of Petal Length')
plt.xlabel('Petal Length')
plt.ylabel('Density')
plt.show()
```

Analysis:
1. Density plot shows the distribution of petal length for all three species.
2. Setosa has a separate distribution with smaller petal lengths.
3. Versicolor has medium petal lengths.
4. Virginica has larger petal lengths.
5. There is overlap between Versicolor and Virginica, while Setosa is clearly distinguishable.

```python
#4. Violin Plot
plt.figure(figsize=(9,6))
sns.violinplot(
    data=df,
    x='species',
    y='petal_length',
    hue='species',
    palette='Set2',
    legend=False
)

plt.title('Violin Plot of Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length')
plt.show()
```

Analysis:
1. Violin plot compares petal length across the three species.
2. Setosa has the smallest petal length distribution.
3. Versicolor has medium petal lengths.
4. Virginica has the largest petal length distribution.
5. The width of each violin represents the density of observations at different values.
6. The plot shows that petal length is a useful feature for distinguishing flower species.

```python
#5. Heatmap(correlation)
plt.figure(figsize=(5,9))
correlation = df.select_dtypes(include='number').corr()
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.9
)

plt.title('Correlation Heatmap of Iris Dataset')
plt.show()
```

Analysis:
1. Petal length and petal width have a strong positive correlation (0.96).
2. This means that as petal length increases, petal width also tends to increase.
3. Sepal length and petal length have a strong positive correlation (approximately 0.87).
4. Sepal width and petal length have a negative correlation (approximately -0.43).
5. Sepal width and petal width also show a negative correlation (approximately -0.37).
6. The diagonal values are always 1 because every variable has a perfect positive correlation with itself.

```python
#6. Joint Plot
sns.jointplot(
    data=df,
    x='sepal_length',
    y='petal_length',
    hue='species',
    kind='scatter',
    height=4
)

plt.suptitle('Joint Plot of Iris Dataset', y=2.08)
plt.show()
```

Analysis:
1. Joint plot shows the relationship between sepal length and petal length.
2. A positive relationship can be observed between the two variables.
3. Setosa flowers have smaller petal lengths compared to the other species.
4. Virginica generally has larger sepal and petal lengths.
5. The marginal plots show the individual distributions of both variables.
6. The visualization helps to understand the relationship between two features and their distributions.

```python
#7. Pair Plot
sns.pairplot(
    df,
    hue='species',
    diag_kind='hist',
    palette='husl'
)

plt.suptitle('Pair Plot of Iris Dataset', y=1.04)
plt.show()
```

Analysis:
1. Pair plot shows relationships between all four numerical features.
2. Petal length and petal width show a strong positive relationship.
3. Setosa is clearly separated from Versicolor and Virginica in petal-related plots.
4. Versicolor and Virginica show some overlap in certain feature combinations.
5. The diagonal graphs represent the individual distributions of each numerical feature.
6. Petal length and petal width are useful features for distinguishing between the three Iris species.
